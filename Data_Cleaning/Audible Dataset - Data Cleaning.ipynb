```python
#Always use copy of your dataframe for data cleaning
audible_df1 = audible.copy()
```

Define:
  1. Separate rating from stars to new column using regex

```python
import re

audible_df1[['Stars' , 'Rating']] = audible_df1['stars'].str.extract(r'(\d+) out of 5 stars(\d+) ratings')

audible_df1['Stars'] = audible_df1['Stars'].astype(float)
audible_df1['Rating'] = audible_df1['Rating'].astype('Int64')


audible_df1.loc[audible_df1['stars'].str.contains("Not rated yet", na =False) , ['Stars' , 'Rating']] = [None , 0]
```

```python
audible_df1
```

Define: 
 1. filling rating null values with zero
 2. change duration in time column to complete min format 
 3. remove writtenby: and narratedby:
 4. can price column type to decimal 


```python
audible_df1['Rating'] = audible_df1['Rating'].fillna(0)
audible_df1['Rating'].isnull().any()
```

```python
def convert_to_minutes(text):
    match = re.search(r'(\d+)\s*hrs?\s*and\s*(\d+)\s*mins?' , text)
    if match:
        hours = int(match.group(1))
        mins = int(match.group(2))
        return hours*60 + mins
    return None

audible_df1['time'] = audible_df1['time'].apply(convert_to_minutes)
```

```python
audible_df1
```

```python
#remove "Writtenby:" from author column
audible_df1['author'] = audible_df1['author'].str.replace(r'Writtenby:','', regex=True) 
```

```python
audible_df1
```

```python
# remove "Narratedby:" from narrator column
audible_df1['narrator'] = audible_df1['narrator'].str.replace(r'Narratedby:', '', regex=True)
audible_df1
```

```python
audible_df1 = audible_df1.rename(columns={'time': 'Time (min)'})
```

```python
audible_df1 = audible_df1.rename(columns={'name': 'Name'})
audible_df1 = audible_df1.rename(columns={'author': 'Author'})
audible_df1 = audible_df1.rename(columns={'narrator': 'Narrator'})
audible_df1 = audible_df1.rename(columns={'releasedate': 'Release Date'})
audible_df1 = audible_df1.rename(columns={'language': 'Language'})
audible_df1 = audible_df1.rename(columns={'price': 'Price'})
audible_df1 = audible_df1.drop(columns=['stars'])
```

```python
audible_df1
```

```python
audible_df1.info()
```

```python
audible_df1['Price'] = audible_df1['Price'].str.replace(',' , '' , regex=False).replace('Free' , '0').astype(float)
```
