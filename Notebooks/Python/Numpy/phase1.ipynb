```python
import numpy as np
```

```python
py_list = [1,2,3] #concat
print("python list multiplication: ",py_list * 2)

np_array = np.array([1,2,3]) #multiply element wise
print("python array multiply: ",np_array * 2)
```

```python
zeros = np.zeros((3,4))
print("zeros array: \n", zeros)

ones = np.ones((2,3))
print("ones array: \n",ones)

full = np.full((2,2) , 5)
print("full array: \n" , full)

random = np.random.random((2,3))
print("random array: \n" , random)

sequence = np.arange(0,10,3) #non inclusive (dont consider boundryb)
print("sequence array: \n",sequence)
```

### Vector , matrix and tensor

```python
vector = np.array([1,2,3])
print("Vector: ",vector)

matrix = np.array([[1,2,3], 
                  [4,5,6]])
print("Matrix: ", matrix)

tensor = np.array([[[1,2] , [3,4]],
                   [[5,6] , [7,8]]]) #multidimensional array in multidimensional arr
print("tensor: ", tensor)

```

### Array properties


```python
arr = np.array([[1,2,3], 
                  [4,5,6]]) 
print("Shape: ",arr.shape)
print("Shape: ",arr.ndim)
print("Shape: ",arr.size)
print("Shape: ",arr.dtype)

```

```python
arr = np.arange(12)
print("original array ", arr)

reshaped = arr.reshape((3,4)) # make flatten arr into col n rows
print("\n Reshaped array ", reshaped)

flattened = reshaped.flatten() # flatten rows and cols
print("\n FLattened array: ", flattened)

array = np.array([[1,2,3], 
                  [4,5,6]]) 

#ravel (return view , instead of copy) both og and ravel point to same memory
raveled = reshaped.ravel()
print("\n raveled array ",raveled)

#transpose
transpose = reshaped.T
print("\n transpose array ", transpose)

```
