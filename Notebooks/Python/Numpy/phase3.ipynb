### Advance operation with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure : [restaurant_id , 2021 , 2022 , 2023 , 2024]
sales_data = np.array([
    [1,250000,180000,220000,100000], #biryani
    [2,120000,140000,160000,190000], #beijing bites
    [3,200000,230000,260000,300000], #pizza hub
    [4,180000,210000,240000,270000], #burger point
    [5,160000,185000,205000,230000]  #chai point
])

print("=== Zomato sales analysis ===")
print("\nSales data shape",sales_data.shape)
print("Sales data dim",sales_data.ndim)
print("Sales data size",sales_data.size)
print("1sr three restaurant", sales_data[:3])
```

```python
#total of yearly data
sum = np.sum(sales_data , axis = 0)
print(sum)
yearly_sale = np.sum(sales_data[: , 1:] , axis =0 )
print("yearly sale total" , yearly_sale)

#min sales per restro
min = np.min(sales_data[: , 1:] , axis = 1)
print("min sale per restro", min)


#max sales per year
max = np.max(sales_data[: , 1:] , axis = 0)
print("max sale per year", max)

#avg sales per restaurant
avg = np.mean(sales_data[: , 1:] , axis = 1)     #axis 0 : col , axis 1 : row
print("avg sale per restro" , avg)

#cumulative sales per 
cumsale = np.cumsum(sales_data[: , 1:] , axis = 1)
print("cum sale\n" , cumsale)

#basically here I'm plotting every years average 
print("plot data",np.mean(cumsale , axis=0))
plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsale , axis=0))
plt.title("Average cumulative sales(total revenue) across all restro")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

### Vector operation

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector addition: ", vector1 + vector2 )
print("Vector multiplication: ", vector1 * vector2 )
print("Vector Dot product: ", np.dot(vector1,vector2) )

angle = np.arccos(np.dot(vector1 , vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("angle " , angle)
```

```python
#Vectorize operations

restro_types = np.array(['briyani' , 'chinese' , 'pizza' , 'burger' , 'cafe'])

vectorized_upper = np.vectorize(str.upper)

result = vectorized_upper(restro_types)
print(result)

## with np.vectorize operation will get performed with each and every ele of array
```

### Broadcasting

Broadcasting is NumPy's way of making arrays with different shapes work together in arithmetic operations.

```python
monthly_avg = sales_data[: , 1:] / 12
print(monthly_avg)

## 12 is broadcast to every ele of array and every ele will get /12
```
