```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1 = np.array([[1,2,3] , [4,5,6]])
array2 = np.random.rand(3,3)
array3 = np.zeros((4,4))

```

#### A .npy file stores exactly one Numpy array , numpy to save and load single arrays.

#### To save multiple arrays use .npz format (which is essentially a ZIP archive of multiple .npy files) or save them individually.

```python
np.save('array1.npy' , array1)
np.save('array2.npy' , array2)
np.save('array3.npy' , array3)
```

```python
loaded_array = np.load('array1.npy') #enter path of npy file
print(loaded_array)
```

```python
try:
    logo = np.load('numpy-logo.npy')

    #display
    plt.figure(figsize=(8,5))
    plt.subplot(121)  # 1 row of subplots , # 2 columns of subplots , 1 for select left subplot (1*2 grid)
    plt.imshow(logo)
    plt.title("Numpy logo")
    plt.grid(False)

except FileNotFoundError:
    print("numpy logo file not found")
```

```python
plt.figure(figsize=(8,5))
plt.subplot(122)
plt.imshow(1-logo)
plt.title("Numpy dark theme logo")
```
