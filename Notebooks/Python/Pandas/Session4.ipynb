## Joins

```python
import pandas as pd

df_users = pd.DataFrame({'user_id': [1, 2, 3], 'name': ['Alice', 'Bob', 'Charlie']})
df_orders = pd.DataFrame({'order_id': [101, 102], 'user_id': [1, 2], 'amount': [250, 450]})
print(df_users)
print(df_orders)
```

```python
## 4 major joins 
# A. Inner join(default)

pd.merge(df_users , df_orders , on='user_id', how='inner')
```

```python
# B. Left join

pd.merge(df_users , df_orders , on='user_id' , how='left')
```

```python
# C. right join
pd.merge(df_users , df_orders , on='user_id' , how='right')
```

```python
# D. outer join
pd.merge(df_users , df_orders , on='user_id' , how='outer')
```

 _Pro-Tip for Messy Data: If your joining columns have different names in the two datasets (e.g., user_id in the left table and customer_id in the right), use left_on and right_on:
pd.merge(df_users, df_orders, left_on='user_id', right_on='customer_id', how='left')_

```python
#SQL Equivalent: Joining two tables on their Primary Keys.
#When to use it: When your DataFrames already have the ID column set as their index.

df_users_idx = df_users.set_index('user_id')
df_orders_idx = df_orders.set_index('user_id')
print(df_users_idx)

result = df_users_idx.join(df_orders_idx , how='inner')
print(result)
```
