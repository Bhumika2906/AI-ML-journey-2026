## DataFrame

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"],
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
# Broadcasting

df['Salary'] = df['Salary'] + 5000
print(df['Salary'])
```

```python
#rename column

df.rename(columns = {"Department": "Dept"} , inplace = True)
df

 
```

```python
#unique data
print(df["Salary"].unique())

#counts
df['Dept'].value_counts()
```

```python
df["Promoted Salary"] = df["Salary"] * 10
df
```

### NULL values( Cleaning data )

```python
df.isnull().sum()
```

```python
df.dropna(how = "all") #if all the values in any row are null then we drop
```

```python
df.dropna(how = "any") #if any values in any row are null then drop that row
# here as not used inplace original df is intect
```

```python
# to fill null value 
df.fillna(0)
```

```python
#to fill null value with mean
#df["Age"].fillna(df["Age"].mean())

df["Salary"].fillna(df["Salary"].median())

df["Salary"].ffill() # forward fill
df["Salary"].bfill() # backward fill

```

```python
df["Name"] = df["Name"].replace('Charlie' , "Rose")
df
```

### Duplicates

```python
df_dup = df[df.duplicated()]
df_dup
```

```python
df_dup = df[df.duplicated(keep = "last")]
df_dup
```

```python
df = df.drop_duplicates()
df
```

```python
#lambda func

df['Promoted Salary'] = df['Promoted Salary'].apply(lambda x : x/10 if x > 650000 else x)
df 
```

```python

df['Name'] = df['Name'].replace('Alice' , 'Alice_fernandes')
df
```

```python
name = "Alice_fernandes"
df['first_name' , 'last_name'] = df["Name"].str.split("_")
df
```

```python
# apply and lambda

def multiplying_salary(x):
    return x*2

df["Salary"] = df["Salary"].apply(multiplying_salary)
df

```
