### Series in python

```python
import pandas as pd
```

```python
s = pd.Series([10,20,30,40,50])
print(s) #print ways
s


```

```python
print(s.dtype)

print(s.values)

print(s.index)

print(s.name)
```

```python
s.name = "numbers"
s
```

```python
#indexing 
print(s[0])
print(s[0:2])
```

```python
#iloc -> location based indexing
print(s.iloc[1])
print(s.iloc[[0,2]])
```

```python
s.name = "calories"
s
```

```python
indx = ["apple" , "banana" ,"grapes" ,"orange" , "berries"]

s.index = indx
print(s)
```

```python
s["grapes"]
```

```python
s.iloc['grapes']
```

```python
# we cannot use labeled index in iloc still need to use s[3(index)] 
#to solve this we use loc -> label based indexing

print(s.loc[['grapes' , 'banana' , 'berries']])

# in label baased indexing start and stop both are included***

print(s['banana': 'orange'])
```

### Conditional selection

```python
fruit_protein = {
    "Avocado": 2.0,       # grams of protein
    "Guava": 2.6,
    "Blackberries": 2.0,
    "Oranges": 0.9,
    "Banana": 1.1,
    "Apples": 0.3,
    "Kiwi": 1.1,
    "Pomegranate": 1.7,
    "Mango": 0.8,
    "Cherries": 1.0
} ## dictionary
```

```python
s2 = pd.Series(fruit_protein , name = "Protein")
s2
```

```python
s2[s2 > 1] 
# if we s2 > 1 it will give masked series of True and False
# passing it in Series it will give only True selected rowss
```

```python
#logical operators : and , or , not

print([(s2 > 0.5 ) & (s2 < 1.0)])

print(s2[((s2 > 0.5 ) & (s2 <= 1.0))])
print(s2[((s2 > 0.5 ) | (s2 <= 1.0))])


```

```python
print("\n" , s2[~(s2 < 1)])
```

```python
s2['Mango'] = 2.8
print(s2)

# to remove any ele

s2 = s2.drop(['Guava' , 'Oranges']) # also can use s.drop(s.index([[1 ,2]]))
print(s2)

```
