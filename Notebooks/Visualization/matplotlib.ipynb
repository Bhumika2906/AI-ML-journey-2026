### 2D Line plot 

- Bivariate analysis
- categorical -> numerical and numerical -> numerical
- use case - Time series data

```python
# plotting a simple function


price = [48000 , 54000 , 57000 , 49000 , 47000 ,  45000]
year = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]

plt.plot(year,price)
```

```python
# from a pandas dataframe
batsman = pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
#plotting multiple plots 
plt.plot(batsman['index']  , batsman['V Kohli'])
plt.plot(batsman['index'] , batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index']  , batsman['V Kohli'])
plt.plot(batsman['index'] , batsman['RG Sharma'])

plt.title('Rohit Sharma vs Virat Kohli Career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#colors and line(width and style) and marker(size)
plt.plot(batsman['index']  , batsman['V Kohli'] , color ="#F19E0F")
plt.plot(batsman['index'] , batsman['RG Sharma'] , color ="#257A31")

plt.title('Rohit Sharma vs Virat Kohli Career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index']  , batsman['V Kohli'] , color ="#F19E0F" , linestyle='solid' , linewidth=3)
plt.plot(batsman['index'] , batsman['RG Sharma'] , color ="#257A31" , linestyle='dashed' , linewidth=2)

plt.title('Rohit Sharma vs Virat Kohli Career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index']  , batsman['V Kohli'] , color ="#F19E0F" , linestyle='solid' , linewidth=3 , marker='D')
plt.plot(batsman['index'] , batsman['RG Sharma'] , color ="#257A31" , linestyle='dashed' , linewidth=2 , marker='o')

plt.title('Rohit Sharma vs Virat Kohli Career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location
plt.plot(batsman['index']  , batsman['V Kohli'] , color ="#F19E0F" , linestyle='solid' , linewidth=3 , marker='D' , label ='Virat')
plt.plot(batsman['index'] , batsman['RG Sharma'] , color ="#257A31" , linestyle='dashed' , linewidth=2 , marker='o' , label='Rohit Sharma')

plt.title('Rohit Sharma vs Virat Kohli Career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc='best')
```

### Scatter plots 
  - Bivariate Analysis ( need two columns to study)
  - numerical vs numerical
  - use case - Finding correlation

```python
#plt.scatter on pandas data
df = pd.read_csv('batter.csv')
df = df.head(50)
df
```

```python
plt.scatter(df['avg'] , df['strike_rate'] , color='green' , marker='*' , s=40)
plt.title('Avg and SR of Top batsman')
plt.xlabel('Average')
plt.ylabel('Strike rate')
```

```python
plt.scatter(df['avg'] , df['strike_rate'] , color='green', s=df['runs']*0.01)
```

```python
#scatter plot using plt.plot
#faster
plt.plot(df['avg'] , df['strike_rate'], 'o' )
```

## Bar chart

  - Bivariate analysis
  - Numerical vs Categorial 
  - Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='pink')
```

```python
plt.barh(colors,children)
```

```python
# color and label
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'] , df['2015'])

```

```python
# to build all three years together
plt.bar(np.arange(df.shape[0]) - 0.2 , df['2015'] , width = 0.2 , color = 'green')
plt.bar(np.arange(df.shape[0]) , df['2016'] , width = 0.2 , color = 'orange')
plt.bar(np.arange(df.shape[0]) + 0.2 , df['2017'] , width = 0.2 , color='indigo')

plt.xticks(np.arange(df.shape[0]) , df['batsman'])
plt.show()
```

```python
np.arange(df.shape[0])
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['redredredredred','blueblueblueblueblue','greengreengreengreeengreen','yellow yellow yellow','pinkpinkpinkpinkpink']

plt.bar(colors , children)
plt.xticks(rotation='vertical')
plt.show() #vertical names print so that htey dont overlap
```

```python
# Stacked Bar chart 
plt.bar(df['batsman'] , df['2017'] , label='2017')
plt.bar(df['batsman'] , df['2016'] , label = '2016' , bottom=df['2017'])
plt.bar(df['batsman'] , df['2015'] , label = '2015' , bottom=df['2016'] + df['2017'])

plt.legend()
plt.show()

#this gives total of 2015 + 2016 + 2017 of each player
```

### Pie chart

  - univariate / bivariate analysis
  - categorial vs numerical
  - use case - to find contribution on a standard scale


```python
#simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data, labels=subjects)

plt.show()
```

```python
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'] , labels=df['batsman'])
plt.show()
```

```python
# percentage and colors

plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct='%0.1f%%')
plt.show()
```

```python
#explode shadow
plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct='%0.1f%%' , explode=[0.5,0,0,0,0,0] , shadow=True)
plt.show()
```
