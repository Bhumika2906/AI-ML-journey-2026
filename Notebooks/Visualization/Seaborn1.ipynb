### 1. Relational Plot
   - to see the  statistical relation between 2 or more variables.
   - Bivariate Analysis
  
  Plots under this section
  - scatterplot
  - lineplot


```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
#scatter plot -> axes level function
sns.scatterplot(data=tips , x ='total_bill' , y='tip' , hue='sex' , style='time' , size='size')
```

```python
# relplot -> figure level -> square shape
sns.relplot(data=tips , x='total_bill' , y='tip' , kind='scatter' , hue='sex' , style='time' , size='size')
```

- We mostly try to use figure plot functions as they are easier to write and have many functions 
- Also figure level is at top and it contain many axes level diff functions in it

```python
# line plot
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
#axes level func
sns.lineplot(data = temp_df , x='year' , y='lifeExp')
```

```python
# using relplot
sns.relplot(data = temp_df , x='year' , y='lifeExp' , kind='line')
```

```python
#hue -> style
temp_df = gap[gap['country'].isin(['India' , 'Brazil' , 'Germany'])]
temp_df
```

```python
sns.relplot(kind = 'line' , data=temp_df , x='year' , y='lifeExp' , hue='country' , style='continent' , size='country')
```

```python
sns.lineplot(data=temp_df , x='year' , y='lifeExp' , hue='country')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot

sns.relplot(data=tips , x='total_bill' , y='tip' , kind='line' , col='sex')
```

```python
sns.relplot(data=tips , x='total_bill' , y='tip' , kind='line' , col='sex' , row='smoker')
```

This col feature is only available in figure level not in axes level

```python
sns.relplot(data=tips , x='total_bill' , y='tip' , kind='line' ,col='day' , col_wrap=2) # 2 col then next row
```

### 2. Matrix plot
 - heatmap
 - clustermap

Heatmap

```python
# Plot rectangular data as a color-encoded matrix
#exist in long format for heatmap need to convert to wide format
temp_df = gap.pivot(index ='country' , columns='year' , values='lifeExp')
print(temp_df)

#axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# annot
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country' , columns='year' , values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df , annot=True , linewidth=0.5 , cmap='summer')
```

Clustermap

```python
# plot a matrix dataset as a hierarchically-clustered heatmap
#this func requires scipy to be available
# use agglomoretive hierarchical clustering
iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[: , [0,1,2,3]])
```
